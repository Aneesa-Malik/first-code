#CODE NO3: Print simple calculate

```python
num=int(input('enter you number'))
num1=int(input('enter you number'))
sign=input('enter the sign')
if(sign=="+"):
    print('The result is:',num+num1)
elif(sign=="-"):
    print('The result is:',num-num1)
elif(sign=="*"):
    print('The result is:',num*num1)
elif(sign=="/"):
    print('The result is:',num/num1)
else:
    print('@#$%^')
```

#CODE NO16:Using nested loop:

```python
a=1
while(a<=5):
    b=0
    while(a>b):
        if(b==a-1):
            print(a,end=' ')
        else:
            print(' ',end=' ')
        b=b+1
    print()
    a=a+1
```
